```python
import matplotlib.pyplot as plt
import numpy as np
import librosa.display

audio, sr = librosa.load("Austin.wav", sr=None)
S_db = 20 * np.log10(abs(librosa.stft(audio)) / np.max(abs(librosa.stft(audio))) + 1e-10)

plt.figure(figsize=(10, 4))
librosa.display.specshow(S_db, sr=sr, x_axis="time", y_axis="log")
plt.colorbar(format="%+2.0f dB")
plt.title("My voice as a spectrogram")
plt.tight_layout()
plt.show()
```

```python
import soundfile as sf

S = abs(librosa.stft(audio))
reconstructed = librosa.griffinlim(S)
sf.write("reconstructed.wav", reconstructed, sr)
print("saved, go listen")
```
